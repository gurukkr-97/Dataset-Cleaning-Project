```python
df["Theory Marks"] = df["Theory Marks"].astype(float)
df["Practical Marks"] = df["Practical Marks"].astype(float)
df["Total Marks"] = df["Total Marks"].astype(float)
df["Attendance Percent"] = df["Attendance Percent"].astype(float)
df["Age"] = df["Age"].astype(int)
```

```python
def zero_count(col_name):
    count = 0
    for i in df.index:
        if df.loc[i,col_name] == 0 or str (df.loc[i,col_name]) == "0":
            count +=1
    print(f"{col_name} zero count : {count}")
```
